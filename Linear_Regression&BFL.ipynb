```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
hour =np.array ([2,4,6,8,10]).reshape (-1,1)
marks = np.array ([40,55,50,78,90])
```

```python
model = LinearRegression ()
model.fit (hour,marks)
```

```python
print ("m:" , model.coef_)
print ("c:" , model.intercept_)
```

```python
plt.scatter (hour,marks,color='blue')
plt.plot (hour,model.predict(hour),color='red',label ="Best Fit Line")
plt.xlabel ("Hours")
plt.ylabel ("Marks")
```

```python
new_hours = np.array([[7]])
predicted_marks = model.predict(new_hours)
print("Predicted marks for 7 hours of study:", predicted_marks[0])
```
