```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression ,SGDRegressor
from sklearn.metrics import mean_squared_error

```

```python
hour =np.array ([2,4,6,8,10]).reshape (-1,1)
marks = np.array ([40,55,50,78,90]) 
hour , marks
```

```python
lin_reg = LinearRegression()
lin_reg.fit (hour,marks)
print ("m:" , lin_reg.coef_[0])
print ("c: " ,lin_reg.intercept_ )
```

```python
#computing mean sqiare value 
predic = lin_reg.predict(hour)
mse= mean_squared_error (marks,predic)
print (mse)
```

```python
#best fit  line 
plt.scatter (hour,marks,color='blue')
plt.plot (hour,predic,color='red',label ="Best Fit Line")
plt.xlabel ("Hours")
plt.ylabel ("Marks")
```

```python
# Using Gradient Descent based Model SGDRegressor
sgd_reg = SGDRegressor(
    max_iter=1000,
    learning_rate="invscaling",
    eta0=0.01,        # corrected
    random_state=42
)

sgd_reg.fit(hour, marks)
pred_SGD = sgd_reg.predict(hour)
mess_SGD = mean_squared_error (marks,pred_SGD)
print(mess_SGD)
```
